```python
import pandas as pd
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.dummy import DummyClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, confusion_matrix
```

```python
df_train = pd.read_csv('../data/kaggle_b2_fraud_train_v3.csv')
df_test = pd.read_csv('../data/kaggle_b2_fraud_test_v3.csv')
```

```python
df_train_clean = df_train.dropna(axis=1)
df_train_clean = df_train_clean.select_dtypes(include=['int64', 'float64', 'bool'])
df_train_clean = df_train_clean.loc[:, df_train_clean.nunique() > 1]
```

```python
X_train = df_train_clean.drop('target_is_fraud', axis=1)
y_train = df_train_clean['target_is_fraud']
```

```python
colonnes_a_garder = X_train.columns.tolist()

# Garder l'index AVANT le nettoyage
df_test_clean = df_test[colonnes_a_garder + ['customer_id']].copy()

# Supprimer les lignes avec des NaN dans les colonnes sélectionnées
df_test_clean = df_test_clean.dropna()

# Séparer customer_id et features
customer_ids = df_test_clean['customer_id']
X_test = df_test_clean[colonnes_a_garder]
```

```python
baseline_model = DummyClassifier(strategy='most_frequent', random_state=42)
baseline_model.fit(X_train, y_train)
y_pred = baseline_model.predict(X_test)
```

```python
predictions_df = pd.DataFrame({
    'customer_id': customer_ids,
    'target': y_pred
})

print(predictions_df.to_string(index=False))
print(f"\nNombre de prédictions: {len(predictions_df)}")
```

```python
predictions_df.to_csv('dummy_predictions.csv', index=False)
```
