```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.tree import DecisionTreeClassifier


df_train = pd.read_csv('../data/kaggle_b2_fraud_train_v3.csv')
df_train
```

```python
threshold = 0.85
df_train = df_train.dropna(axis=1, thresh=len(df_train) * (1 - threshold))
```

```python
df_train = df_train.dropna()
```

```python
le = LabelEncoder()
for col in df_train.select_dtypes(include=['object']).columns:
    df_train[col] = le.fit_transform(df_train[col])
```

```python
X = df_train.drop('target_is_fraud', axis=1)
y = df_train['target_is_fraud']
```

```python
model = DecisionTreeClassifier(random_state=42)
model.fit(X, y)
```

```python
print(f"Colonnes avant : {len(df_train.columns)}")
print(f"Lignes après nettoyage : {len(df_train)}")
```

```python
df_train.to_csv('dataset_clean.csv', index=False)
```
